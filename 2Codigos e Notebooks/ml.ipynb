```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.read_csv('../Data/Gold/games_dataset_cleaned.csv')

colunas_removidas = ['appid', 'name', 'positive', 'tags', 'negative', 'userscore', 'average_forever','median_2weeks', 'median_forever', 'initialprice', 'developer', 'publisher', 'discount', 'total_reviews', 'average_2weeks', 'avg_playtime_hours', 'published_store', 'published_store']
df = df.drop(colunas_removidas, axis=1)
# print(colunas_removidas)
df.head()
```

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
```

```python
numeric_cols.remove('price')
categorical_cols.append('owners')
```

```python
print(numeric_cols)
print(categorical_cols)
```

```python
X = df.drop('price', axis=1)
y = df['price']
```

```python
#criando o pipeline de pré-processamento
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())])
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))])

# aplicando pipeline nos dados
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_cols),
        ('cat', categorical_transformer, categorical_cols)])

X = preprocessor.fit_transform(X)
```

```python
# split dos dados
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=295)
```

```python
# criando o modelo
model = SVR()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print("Predictions:", y_pred)
```

```python
# avaliando a acurácia
accuracy = model.score(X_test, y_test)
print("Accuracy:", accuracy)
```
